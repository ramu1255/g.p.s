```python
# ============================================
# GPS TRILATERATION SIMULATOR
# Cell 1: Imports
# ============================================

import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
from matplotlib.patches import Circle
```

```python
# ============================================
# Cell 2: Problem Inputs
# ============================================

# Known tower positions
tower1 = (0, 0)
tower2 = (10, 0)
tower3 = (4, 8)

# Benchmark phone position used to generate
# consistent distances:
# Actual hidden position = (5, 3)

# Distances calculated using the Euclidean distance formula
d1 = np.sqrt((5 - 0)**2 + (3 - 0)**2)
d2 = np.sqrt((5 - 10)**2 + (3 - 0)**2)
d3 = np.sqrt((5 - 4)**2 + (3 - 8)**2)

T1 = np.array(tower1, dtype=float)
T2 = np.array(tower2, dtype=float)
T3 = np.array(tower3, dtype=float)

distances = np.array([d1, d2, d3], dtype=float)

print("Tower Coordinates")
print("-----------------")
print(f"Tower 1: {T1}")
print(f"Tower 2: {T2}")
print(f"Tower 3: {T3}")

print("\nMeasured Distances")
print("------------------")
print(f"d1 = {d1:.6f}")
print(f"d2 = {d2:.6f}")
print(f"d3 = {d3:.6f}")

print("\nBenchmark hidden position: (5, 3)")
```

```python
# ============================================
# Cell 3: Mathematical Model
# ============================================

x, y = sp.symbols('x y')

# Convert coordinates to SymPy values
x1, y1 = map(sp.Float, T1)
x2, y2 = map(sp.Float, T2)
x3, y3 = map(sp.Float, T3)

D1, D2, D3 = map(sp.Float, distances)

# Three circle equations
eq1 = (x - x1)**2 + (y - y1)**2 - D1**2
eq2 = (x - x2)**2 + (y - y2)**2 - D2**2
eq3 = (x - x3)**2 + (y - y3)**2 - D3**2

print("Circle Equations")
print("================")

print("Tower 1:")
sp.pprint(sp.Eq((x - x1)**2 + (y - y1)**2, D1**2))

print("\nTower 2:")
sp.pprint(sp.Eq((x - x2)**2 + (y - y2)**2, D2**2))

print("\nTower 3:")
sp.pprint(sp.Eq((x - x3)**2 + (y - y3)**2, D3**2))
```

```python
# ============================================
# Cell 4: Linearization
# ============================================

# Subtract pairs of circle equations
linear_expr_1 = sp.expand(eq2 - eq1)
linear_expr_2 = sp.expand(eq3 - eq1)

print("Linear Equations After Subtraction")
print("==================================")

print("\nEquation 1:")
sp.pprint(sp.Eq(linear_expr_1, 0))

print("\nEquation 2:")
sp.pprint(sp.Eq(linear_expr_2, 0))
```

```python
# ============================================
# Cell 5: Extract A, B, C, D, E, F
# ============================================

# Put each equation into:
# A*x + B*y = C

poly1 = sp.Poly(linear_expr_1, x, y)
poly2 = sp.Poly(linear_expr_2, x, y)

A = float(poly1.coeff_monomial(x))
B = float(poly1.coeff_monomial(y))
C = float(-poly1.coeff_monomial(1))

D = float(poly2.coeff_monomial(x))
E = float(poly2.coeff_monomial(y))
F = float(-poly2.coeff_monomial(1))

print("System of Linear Equations")
print("==========================")

print(f"{A:.6f}x + {B:.6f}y = {C:.6f}")
print(f"{D:.6f}x + {E:.6f}y = {F:.6f}")
```

```python
# ============================================
# Cell 6: Math Engine
# ============================================

# Coefficient matrix
M = np.array([
    [A, B],
    [D, E]
], dtype=float)

# Right-hand side
b = np.array([C, F], dtype=float)

# Check whether the system is solvable
determinant = np.linalg.det(M)

if abs(determinant) < 1e-10:
    raise ValueError(
        "The linear system cannot be solved reliably. "
        "The tower geometry may be degenerate."
    )

# Solve for x and y
solution = np.linalg.solve(M, b)

phone_x, phone_y = solution

print("Calculated Phone Location")
print("=========================")
print(f"x = {phone_x:.6f}")
print(f"y = {phone_y:.6f}")
```

```python
# ============================================
# Cell 7: Verification
# ============================================

phone = np.array([phone_x, phone_y])

calculated_d1 = np.linalg.norm(phone - T1)
calculated_d2 = np.linalg.norm(phone - T2)
calculated_d3 = np.linalg.norm(phone - T3)

print("Verification")
print("============")

print(f"{'Tower':<10}{'Input Distance':<20}{'Calculated Distance':<20}")

print(
    f"{'Tower 1':<10}"
    f"{d1:<20.6f}"
    f"{calculated_d1:<20.6f}"
)

print(
    f"{'Tower 2':<10}"
    f"{d2:<20.6f}"
    f"{calculated_d2:<20.6f}"
)

print(
    f"{'Tower 3':<10}"
    f"{d3:<20.6f}"
    f"{calculated_d3:<20.6f}"
)
```

```python
# ============================================
# Cell 8: Output Display
# ============================================

print("=" * 50)
print("       GPS TRILATERATION RESULT")
print("=" * 50)

print(f"\nEstimated Phone Location:")
print(f"    x = {phone_x:.4f}")
print(f"    y = {phone_y:.4f}")

print("\nMathematical Model:")
print("    Three circle equations")
print("          ↓")
print("    Subtract pairs")
print("          ↓")
print("    Two linear equations")
print("          ↓")
print("    Solve Ax + By = C")
print("          ↓")
print("    Estimated (x, y)")

print("\nStatus: Location calculated successfully.")
```

```python
# ============================================
# CELL 1: IMPORTS
# ============================================

import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
from matplotlib.patches import Circle
```

```python
# ============================================
# CELL 2: INPUTS
# ============================================

# Tower coordinates
T1 = np.array([0.0, 0.0])
T2 = np.array([10.0, 0.0])
T3 = np.array([4.0, 8.0])

# Benchmark phone location
# This is used only to generate test distances.
# The algorithm will calculate the location independently.
hidden_phone = np.array([5.0, 3.0])

# Distances from towers to phone
d1 = np.linalg.norm(hidden_phone - T1)
d2 = np.linalg.norm(hidden_phone - T2)
d3 = np.linalg.norm(hidden_phone - T3)

distances = np.array([d1, d2, d3])

print("INPUT DATA")
print("-" * 40)

print(f"Tower 1: {T1}")
print(f"Tower 2: {T2}")
print(f"Tower 3: {T3}")

print("\nDistances:")
print(f"d1 = {d1:.6f}")
print(f"d2 = {d2:.6f}")
print(f"d3 = {d3:.6f}")
```

```python
# ============================================
# CELL 3: MATHEMATICAL MODEL
# ============================================

x, y = sp.symbols('x y')

x1, y1 = T1
x2, y2 = T2
x3, y3 = T3

# Circle equations
eq1 = (x - x1)**2 + (y - y1)**2 - d1**2
eq2 = (x - x2)**2 + (y - y2)**2 - d2**2
eq3 = (x - x3)**2 + (y - y3)**2 - d3**2

print("THREE CIRCLE EQUATIONS")
print("=" * 50)

print("\nTower 1:")
sp.pprint(sp.Eq(
    (x - x1)**2 + (y - y1)**2,
    d1**2
))

print("\nTower 2:")
sp.pprint(sp.Eq(
    (x - x2)**2 + (y - y2)**2,
    d2**2
))

print("\nTower 3:")
sp.pprint(sp.Eq(
    (x - x3)**2 + (y - y3)**2,
    d3**2
))
```

```python
# ============================================
# CELL 4: LINEARIZATION
# ============================================

# Subtract Circle 1 from Circle 2
linear1 = sp.expand(eq2 - eq1)

# Subtract Circle 1 from Circle 3
linear2 = sp.expand(eq3 - eq1)

print("LINEAR EQUATIONS")
print("=" * 50)

print("\nEquation 1:")
sp.pprint(sp.Eq(linear1, 0))

print("\nEquation 2:")
sp.pprint(sp.Eq(linear2, 0))

print("\nThe x² and y² terms have cancelled.")
```

```python
# ============================================
# CELL 5: LINEAR SYSTEM
# ============================================

poly1 = sp.Poly(linear1, x, y)
poly2 = sp.Poly(linear2, x, y)

A = float(poly1.coeff_monomial(x))
B = float(poly1.coeff_monomial(y))
C = float(-poly1.coeff_monomial(1))

D = float(poly2.coeff_monomial(x))
E = float(poly2.coeff_monomial(y))
F = float(-poly2.coeff_monomial(1))

print("SYSTEM OF LINEAR EQUATIONS")
print("=" * 50)

print(f"{A:.6f}x + {B:.6f}y = {C:.6f}")
print(f"{D:.6f}x + {E:.6f}y = {F:.6f}")

# Matrix form
M = np.array([
    [A, B],
    [D, E]
])

b = np.array([
    C,
    F
])

print("\nMatrix A:")
print(M)

print("\nMatrix b:")
print(b)
```

```python
# ============================================
# CELL 6: MATH ENGINE
# ============================================

# Check determinant
determinant = np.linalg.det(M)

if abs(determinant) < 1e-10:
    raise ValueError(
        "No unique solution. "
        "Choose better tower positions."
    )

# Solve:
# M [x y]^T = b

solution = np.linalg.solve(M, b)

phone_x = solution[0]
phone_y = solution[1]

phone_location = np.array([
    phone_x,
    phone_y
])

print("CALCULATED LOCATION")
print("=" * 50)

print(f"x = {phone_x:.6f}")
print(f"y = {phone_y:.6f}")

print(f"\nPhone Location = ({phone_x:.6f}, {phone_y:.6f})")
```

```python
# ============================================
# CELL 7: VERIFICATION
# ============================================

# Calculate distances again using calculated location

calculated_d1 = np.linalg.norm(phone_location - T1)
calculated_d2 = np.linalg.norm(phone_location - T2)
calculated_d3 = np.linalg.norm(phone_location - T3)

print("VERIFICATION")
print("=" * 65)

print(
    f"{'Tower':<10}"
    f"{'Input Distance':<20}"
    f"{'Calculated Distance':<20}"
)

print(
    f"{'Tower 1':<10}"
    f"{d1:<20.6f}"
    f"{calculated_d1:<20.6f}"
)

print(
    f"{'Tower 2':<10}"
    f"{d2:<20.6f}"
    f"{calculated_d2:<20.6f}"
)

print(
    f"{'Tower 3':<10}"
    f"{d3:<20.6f}"
    f"{calculated_d3:<20.6f}"
)
```

```python
# ============================================
# CELL 8: FINAL OUTPUT
# ============================================

print("=" * 60)
print("       GPS TRILATERATION RESULT")
print("=" * 60)

print("\nEstimated Phone Location:")
print(f"X = {phone_x:.4f}")
print(f"Y = {phone_y:.4f}")

print("\nMathematical Process:")
print("3 distances")
print("      ↓")
print("3 circle equations")
print("      ↓")
print("Subtract pairs")
print("      ↓")
print("2 linear equations")
print("      ↓")
print("Solve linear system")
print("      ↓")
print("(x, y) = Phone Location")
```
